```python
from datetime import datetime, timedelta

# Definimos una clase para el agente
class AgentePersonal:

    def __init__(self):
        self.actividades = []

    # Agregar actividad
    def agregar_actividad(self, nombre, tipo, prioridad, duracion):
        actividad = {
            "nombre": nombre,
            "tipo": tipo,  # estudio, gym, descanso
            "prioridad": prioridad,  # 1 alta, 2 media, 3 baja
            "duracion": duracion  # en horas
        }
        self.actividades.append(actividad)

    # Ordenar por prioridad
    def organizar_actividades(self):
        self.actividades.sort(key=lambda x: x["prioridad"])

    # Generar horario automático
    def generar_horario(self, hora_inicio="08:00"):
        self.organizar_actividades()
        horario = []

        hora_actual = datetime.strptime(hora_inicio, "%H:%M")

        for act in self.actividades:
            inicio = hora_actual
            fin = inicio + timedelta(hours=act["duracion"])

            horario.append({
                "actividad": act["nombre"],
                "tipo": act["tipo"],
                "inicio": inicio.strftime("%H:%M"),
                "fin": fin.strftime("%H:%M")
            })

            hora_actual = fin

        return horario

    # Mostrar horario
    def mostrar_horario(self, horario):
        print("\n--- PLAN DEL DÍA ---")
        for bloque in horario:
            print(f"{bloque['inicio']} - {bloque['fin']} | {bloque['actividad']} ({bloque['tipo']})")



# USO DEL AGENTE


agente = AgentePersonal()

# Agregamos actividades
agente.agregar_actividad("Estudiar parcial IA", "estudio", 1, 2)
agente.agregar_actividad("Hacer tareas", "estudio", 2, 2)
agente.agregar_actividad("Ir al gimnasio", "gym", 2, 1)
agente.agregar_actividad("Descanso", "descanso", 3, 1)

# Generar horario
horario = agente.generar_horario()

# Mostrar resultado
agente.mostrar_horario(horario)
```
