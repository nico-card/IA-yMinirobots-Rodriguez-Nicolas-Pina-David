```python
from collections import Counter
import math

class SistemaRAG:

    def __init__(self, documentos):
        self.documentos = documentos

    # Convertir texto a vector (tipo embedding simple)
    def vectorizar(self, texto):
        palabras = texto.lower().split()
        return Counter(palabras)

    # Similitud coseno
    def similitud(self, v1, v2):
        interseccion = set(v1.keys()) & set(v2.keys())
        numerador = sum(v1[x] * v2[x] for x in interseccion)

        suma1 = sum(v1[x]**2 for x in v1)
        suma2 = sum(v2[x]**2 for x in v2)

        denominador = math.sqrt(suma1) * math.sqrt(suma2)

        if denominador == 0:
            return 0
        else:
            return numerador / denominador

    # Buscar mejor documento
    def recuperar(self, pregunta):
        vector_pregunta = self.vectorizar(pregunta)

        mejor_doc = None
        mejor_score = 0

        for doc in self.documentos:
            vector_doc = self.vectorizar(doc)
            score = self.similitud(vector_pregunta, vector_doc)

            if score > mejor_score:
                mejor_score = score
                mejor_doc = doc

        return mejor_doc

    # Generar respuesta
    def responder(self, pregunta):
        contexto = self.recuperar(pregunta)

        if contexto:
            return f"Según la Universidad Nacional de Colombia: {contexto}"
        else:
            return "No se encontró información relevante."


# DATOS DE LA UNIVERSIDAD


documentos = [
    "El calendario académico inicia en febrero en la Universidad Nacional de Colombia.",
    "La biblioteca abre de lunes a viernes de 8am a 8pm.",
    "El proceso de matrícula se realiza en enero y julio.",
    "La Universidad Nacional de Colombia tiene sedes en varias ciudades del país.",
    "El bienestar universitario ofrece servicios a los estudiantes."
]


# USO


rag = SistemaRAG(documentos)

print(rag.responder("¿cuándo se realiza el proceso de matrícula?"))
print(rag.responder("horario de biblioteca"))
print(rag.responder("sedes de la universidad"))
```
