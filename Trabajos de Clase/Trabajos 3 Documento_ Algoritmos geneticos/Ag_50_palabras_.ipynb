```python
import random
import string
from gtts import gTTS
from IPython.display import Audio

#PALABRA OBJETIVO
objetivo = "sol"

#Parámetros
tam_poblacion = 50
tasa_mutacion = 0.1
genes = string.ascii_lowercase + " "

#FUNCIONES DEL AG
#palabra aleatoria
def palabra_aleatoria():
    return ''.join(random.choice(genes) for _ in range(len(objetivo)))

#población inicial
def crear_poblacion():
    return [palabra_aleatoria() for _ in range(tam_poblacion)]

#cuántas letras coinciden
def aptitud(palabra):
    return sum(1 for i in range(len(objetivo)) if palabra[i] == objetivo[i])

# Selección de los mejores
def seleccionar(poblacion):
    return sorted(poblacion, key=lambda x: aptitud(x), reverse=True)[:10]

# Cruce
def cruzar(p1, p2):
    punto = random.randint(0, len(objetivo)-1)
    return p1[:punto] + p2[punto:]

# Mutación
def mutar(palabra):
    palabra = list(palabra)
    for i in range(len(palabra)):
        if random.random() < tasa_mutacion:
            palabra[i] = random.choice(genes)
    return ''.join(palabra)

# ALGORITMO GENÉTICO
poblacion = crear_poblacion()
generacion = 0

while True:
    poblacion = sorted(poblacion, key=lambda x: aptitud(x), reverse=True)
    mejor = poblacion[0]

    print(f"Generación {generacion}: {mejor}")

    # Si llegó a la palabra objetivo
    if mejor == objetivo:
        print("🎉 Encontrada la palabra objetivo!")
        break

    seleccionados = seleccionar(poblacion)

    nueva_poblacion = []
    while len(nueva_poblacion) < tam_poblacion:
        p1, p2 = random.sample(seleccionados, 2)
        hijo = cruzar(p1, p2)
        hijo = mutar(hijo)
        nueva_poblacion.append(hijo)

    poblacion = nueva_poblacion
    generacion += 1

# VOZ
tts = gTTS(mejor, lang="es")
tts.save("resultado.mp3")

Audio("resultado.mp3")
```
