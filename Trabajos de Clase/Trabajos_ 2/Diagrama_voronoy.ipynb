## Introducción

Para este análisis se tomó como base un mapa de la ciudad de Zipaquirá (Cundinamarca) y se aplicaron diagramas de Voronoi sobre tres tipos de servicios: **centros de salud (hospitales)**, **centros educativos (colegios)** y **droguerías**. El objetivo es visualizar la distribución espacial de estos equipamientos e identificar posibles zonas desatendidas.

## 1. Centros de Salud (Hospitales)

### Diagrama de Voronoi para hospitales

![Hospitales - Voronoi](
)

**Análisis:**

La distribución de los centros de salud muestra una fuerte concentración en el centro de Zipaquirá, mientras que las zonas periféricas, como el barrio **El Carmen** y la zona oriental del mapa, carecen por completo de un hospital cercano. Únicamente se observa **Arcasalud** como centro de salud relevante en el centro, dejando un vacío importante en el oriente y norte de la ciudad. Esta situación genera dificultades de acceso a servicios de salud para los habitantes de esos sectores.

## 2. Centros Educativos (Colegios)

### Diagrama de Voronoi para colegios

![Colegios - Voronoi](
)

**Análisis:**

La oferta educativa en Zipaquirá está mejor distribuida que la de salud. Aunque también existe una mayor densidad en el centro, se aprecian colegios en algunas áreas periféricas. Sin embargo, en el **norte**, **oriente** y **nororiente** la cobertura podría mejorarse; sería recomendable la construcción de al menos dos colegios adicionales en el norte para atender adecuadamente a la población de esos sectores. En términos generales, la situación educativa es aceptable, pero existen oportunidades de mejora en la periferia.

## 3. Droguerías

### Diagrama de Voronoi para droguerías

![Droguerías - Voronoi](
)

**Análisis:**

Las droguerías presentan una alta concentración en el centro y centro‑occidente, mientras que en el oriente la densidad disminuye drásticamente. Zonas como **El Carmen**, **La Lomita**, **El Recreo**, **La Paz** (que solo cuenta con una droguería) y la zona sur del mapa carecen de una droguería cercana. También se observan vacíos en algunos tramos de la **calle 8**. Esta situación evidencia una necesidad urgente de establecer más droguerías en esas áreas para garantizar el acceso a medicamentos.

## Conclusiones y Preguntas

### Pregunta 1: ¿Considera que puede faltar una droguería, un centro de atención de salud o un colegio?

* **Droguerías:** Sí, hacen falta en varias zonas. Por ejemplo, en **La Paz** solo hay una, lo cual es insuficiente; también en sectores como **La Lomita**, **El Recreo**, la zona sur del mapa, y en algunas partes de la **calle 8**. La cobertura es deficiente en el oriente y sur.
* **Colegios:** En general están bien distribuidos, pero podría mejorarse en el **norte** con dos colegios adicionales, y en el **oriente y nororiente** también se requieren más instituciones para cubrir adecuadamente esas áreas.
* **Hospitales:** Definitivamente hacen falta. Zonas como **El Carmen** y la zona oriental del mapa no tienen centros de salud cercanos; solo se observa **Arcasalud** en el centro, lo que deja a gran parte de la población sin acceso rápido a servicios de salud.

### Pregunta 2: ¿Hay alguna relación entre los diagramas?

Sí, en todos los diagramas (colegios, droguerías, hospitales) se observa una alta concentración en el **centro y centro‑occidente** de Zipaquirá, mientras que hacia el **oriente, norte y sur** la densidad disminuye significativamente. Esto revela un patrón de **centralización de los servicios**, lo que puede generar desigualdades en el acceso para los habitantes de las zonas periféricas. Especialmente crítico es el caso de los hospitales y droguerías, donde la falta de cobertura es más evidente.
