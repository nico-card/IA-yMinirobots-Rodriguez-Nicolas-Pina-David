```python
import operator
import random
from deap import base, creator, tools, gp, algorithms

# Definir conjunto de funciones
pset = gp.PrimitiveSet("MAIN", 1)

pset.addPrimitive(operator.add, 2)
pset.addPrimitive(operator.sub, 2)
pset.addPrimitive(operator.mul, 2)

def protectedDiv(a, b):
    if b == 0:
        return 1
    return a / b

pset.addPrimitive(protectedDiv, 2)

# Variable de entrada
pset.renameArguments(ARG0='error')

# Crear fitness
creator.create("FitnessMax", base.Fitness, weights=(1.0,))
creator.create("Individual", gp.PrimitiveTree, fitness=creator.FitnessMax)

toolbox = base.Toolbox()

toolbox.register("expr", gp.genHalfAndHalf, pset=pset, min_=1, max_=3)
toolbox.register("individual", tools.initIterate, creator.Individual, toolbox.expr)
toolbox.register("population", tools.initRepeat, list, toolbox.individual)

# Función de evaluación
def evaluar(individual):
    func = gp.compile(individual, pset)

    temperatura = 18
    deseada = 22
    error_total = 0

    for i in range(20):

        error = deseada - temperatura

        control = func(error)

        temperatura += control * 0.1

        error_total += abs(deseada - temperatura)

    fitness = 1 / (0.1 + error_total)

    return (fitness,)

toolbox.register("evaluate", evaluar)
toolbox.register("mate", gp.cxOnePoint)
toolbox.register("mutate", gp.mutUniform, expr=toolbox.expr, pset=pset)
toolbox.register("select", tools.selTournament, tournsize=3)

# Crear población
pop = toolbox.population(n=50)

# Ejecutar algoritmo
algorithms.eaSimple(pop, toolbox, 0.7, 0.1, 20)

# Mejor solución
best = tools.selBest(pop, 1)[0]

print("Mejor controlador encontrado:")
print(best)
```
